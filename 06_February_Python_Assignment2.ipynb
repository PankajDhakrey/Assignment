```python
##42. Encode a string using a Caesar cipher.

alphabet=["a","b","c","d","e","f","g","h","i","j","k","l","m","n","o","p","q","r","s","t","u","v","w","x","y","z"]

s="Hello pankaj"

st=s.lower()
encoded_text=""
position_shift=10

for i in st:
  if i in alphabet:
    position=alphabet.index(i)
    new_position=position+position_shift
    encoded_text+=alphabet[new_position]
  else:
    encoded_text+=i
print(encoded_text)
```

```python
#43. Decode a Caesar cipher encoded string.
text = ("rovvy zkxukt")
shift = int(10)

choice =("Decode").strip().lower()

if choice == 'decode':
    shift = -shift

result_text = ""

for char in text:
    if char.isalpha():
        offset = 65 if char.isupper() else 97
        result_char = chr((ord(char) - offset + shift) % 26 + offset)
        result_text += result_char
    else:
        result_text += char

print(result_text)
```

```python
#32. Create a tuple of the first 5 prime numbers.
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

prime_numbers = []
num = 2

for _ in range(5):
    while not is_prime(num):
        num += 1
    prime_numbers.append(num)
    num += 1

    prime_tuple = tuple(prime_numbers)

print(prime_tuple)
```
